```python
# Step 3 — Separation chalain (vocals + accompaniment)
import glob, os
songs = [f for f in glob.glob('*.mp3')+glob.glob('*.wav')+glob.glob('*.m4a')+glob.glob('*.flac') if not f.startswith('SurGeet')]
assert songs, 'Koi audio file nahi mili — pehle Step 2 se song upload karein'
song = songs[0]
print('Song:', song)
!python -m demucs --two-stems=vocals -o separated "{song}"
print('Separation mukammal!')
```

```python
# Step 5 — Stems download karein
import glob
from google.colab import files
for f in sorted(glob.glob('separated/**/*.wav', recursive=True)):
    print('Downloading:', f)
    files.download(f)
```
